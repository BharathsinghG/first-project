```python
# Step 1: Install required packages
!pip install pandas scikit-learn

# Step 2: Import libraries
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import matplotlib.pyplot as plt
from google.colab import files
from io import StringIO
```

```python
# Step 3: Upload the provided 'transactions.csv' file or skip if it already exists
import os
if not os.path.exists('transactions.csv'):
  uploaded = files.upload()  # Remove filename to allow uploading any file
else:
  print("File 'transactions.csv' already exists. Skipping upload.")
```

```python
# Step 4: Read the CSV into DataFrame
df = pd.read_csv("transactions.csv")
df.head()
```

```python
# Step 5: Preprocess data
df.columns = [col.strip() for col in df.columns]
df = df[['Date', 'Description', 'Amount']]
df['Amount'] = df['Amount'].astype(float)
df.head()
```

```python
# Step 6: Simple Expense Categorization using Naive Bayes

# Sample training data
training_data = {
    'Description': [
        'Dominos Pizza', 'Amazon Purchase', 'Uber Ride', 'Netflix Subscription',
        'Electric Bill', 'Salary Payment', 'Grocery Store', 'Insurance Premium',
        'Cafe Coffee Day', 'Gas Station', 'Book Store'
    ],
    'Category': [
        'Food', 'Shopping', 'Transport', 'Entertainment',
        'Utilities', 'Income', 'Groceries', 'Insurance',
        'Food', 'Transport', 'Shopping'
    ]
}

train_df = pd.DataFrame(training_data)

# Train model
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(train_df['Description'])
y_train = train_df['Category']

model = MultinomialNB()
model.fit(X_train, y_train)

# Predict categories
X_test = vectorizer.transform(df['Description'])
df['Category'] = model.predict(X_test)
df.head()
```

```python
# Step 7: Analyze and visualize expenses
summary = df[df['Amount'] < 0].groupby('Category')['Amount'].sum().sort_values().abs()
summary.plot(kind='bar', title='Expenses by Category', figsize=(10, 5), color='skyblue')
plt.ylabel('Amount (₹)')
plt.grid(True)
plt.show()
```

```python
# Step 8: Financial Insights
total_income = df[df['Amount'] > 0]['Amount'].sum()
total_expenses = abs(df[df['Amount'] < 0]['Amount'].sum()) # Use abs() function here
savings_rate = (total_income - total_expenses) / total_income * 100

print(f"💰 Total Income: ₹{total_income:.2f}")
print(f"💸 Total Expenses: ₹{total_expenses:.2f}")
print(f"📊 Savings Rate: {savings_rate:.1f}%")

if savings_rate < 20:
    print("⚠️ Try reducing non-essential expenses to save more.")
elif savings_rate >= 50:
    print("✅ Excellent! You're saving a high percentage of your income.")
else:
    print("👍 Good job. You can optimize further for better savings.")
```
